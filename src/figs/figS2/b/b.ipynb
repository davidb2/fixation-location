# Figure S2b
Mean number of head advances divided by $N-1$ against the mutant fitness, for $N=10$, $20$, $40$, and $80$, with the drift value $r/(r-1)$.

```python
import json

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import LinearSegmentedColormap

from fixloc import style as S

d = json.load(open('data.json'))
```

```python
CM_N = LinearSegmentedColormap.from_list('N', [S.LRED, S.RED, (0.25, 0.04, 0.04)])
rr = np.array(d['r'])
fig = plt.figure(figsize=(7.0, 5.4))
ax = fig.add_axes([0.075, 0.085, 0.235, 0.3])
for k, N in enumerate(d['N']):
  ax.plot(rr, d['mean_over_Nm1'][str(N)], color=CM_N(0.18 + 0.27 * k), lw=1.1, label=r'$N=%d$' % N)
ax.plot(rr, d['drift'], color=S.INK, lw=0.9, ls=(0, (3, 2)), label=r'Drift, $r/(r-1)$')
ax.set_xlim(1, 6)
ax.set_ylim(1, 8)
ax.set_xticks([1, 2, 3, 4, 5, 6])
ax.set_xlabel(r'Mutant fitness, $r$')
ax.set_ylabel(r'Scaled mean, $\mathbb{E}[H]/(N-1)$')
S.legend(ax, loc='upper right', fontsize=6.3, handlelength=1.6)
S.save_panel(fig, 'b')
```
