# Figure S2
Assembles the panels a to d and adds the legend shared by panels c and d.

```python
import json

from matplotlib.colors import LinearSegmentedColormap
from matplotlib.lines import Line2D

from fixloc import style as S

rs = json.load(open('c/data.json'))['rs']
CM_R = LinearSegmentedColormap.from_list('r', [S.LBLUE, S.BLUE, (0.03, 0.15, 0.35)])
RCOL = {r: CM_R(x) for r, x in zip(rs, (0.2, 0.47, 0.74, 1.0))}
fig = S.assemble((7.0, 5.4), ['a/a.png', 'b/b.png', 'c/c.png', 'd/d.png'])
hs = [Line2D([], [], color=RCOL[r], lw=1.2, label=r'$r=%g$' % r) for r in rs]
hs += [
  Line2D([], [], color=S.INK, ls='', marker='o', ms=2.6, mec='white', mew=0.3, label='Exact recursion'),
  Line2D([], [], color=S.INK, lw=1.1, label='Theorem'),
  Line2D([], [], color=S.INK, lw=0.9, ls=(0, (3, 2)), label='Drift prediction'),
]
lg = fig.legend(
  handles=hs,
  loc='upper center',
  bbox_to_anchor=(0.688, 0.525),
  ncol=4,
  fontsize=6.4,
  handlelength=1.8,
  columnspacing=1.2,
)
lg.get_frame().set_linewidth(0.5)
S.letter(fig, 0.004, 0.985, 'a')
S.letter(fig, 0.004, 0.43, 'b')
S.letter(fig, 0.335, 0.43, 'c')
S.letter(fig, 0.665, 0.43, 'd')
S.save_figure(fig, 'figS2')
```
