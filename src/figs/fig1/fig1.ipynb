# Figure 1
Assembles the panels a to d.

```python
from fixloc import style as S

H0, H = 3.1, 3.22  # panel a adds a strip of 0.12 inch on top for the time arrow; panels b, c, d keep their offsets
fig = S.assemble((7.0, H), ['a/a.png', 'b/b.png', 'c/c.png', 'd/d.png'])
S.letter(fig, 0.004, 1 - 0.031 / H, 'a')
S.letter(fig, 0.004, 0.585 * H0 / H, 'b')
S.letter(fig, 0.53, 0.585 * H0 / H, 'c')
S.letter(fig, 0.76, 0.585 * H0 / H, 'd')
S.save_figure(fig, 'fig1')
```
