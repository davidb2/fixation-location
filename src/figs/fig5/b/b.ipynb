# Figure 5b
Probability $p_1$ of a fixation location on island $1$ against the mutant fitness $r$, for $N=2$ to $30$, from one mutant on island $1$, with $m_{12}=10^{-3}$ and $m_{21}=10^{-1}$.

```python
import json

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import LinearSegmentedColormap

from fixloc import style as S

d = json.load(open('data.json'))
rs = np.array(d['r'])
CM_N = LinearSegmentedColormap.from_list('N', [S.LRED, S.RED, (0.25, 0.04, 0.04)])
```

```python
fig = plt.figure(figsize=(7.0, 6.2))
ax = fig.add_axes([0.6, 0.705, 0.29, 0.255])
ax.axhspan(0.5, 1, color=S.LGREEN, alpha=0.45, lw=0, zorder=0)
ax.axhspan(0, 0.5, color=S.LPURPLE, alpha=0.45, lw=0, zorder=0)
for N in range(2, 31, 2):
  ax.plot(rs, d['p1'][str(N)], color=CM_N((N - 2) / 28), lw=1.0, zorder=2)
ax.set_xlim(1, 1.5)
ax.set_ylim(0, 0.85)
ax.set_xticks([1.0, 1.1, 1.2, 1.3, 1.4, 1.5])
ax.set_yticks([0, 0.2, 0.4, 0.6, 0.8])
ax.set_xlabel(r'Mutant fitness, $r$')
ax.set_ylabel('Probability of a fixation location\non island $1$, $p_1$')
cax = fig.add_axes([0.905, 0.73, 0.009, 0.2])
S.colorbar(fig, cax, CM_N, 2, 30, [2, 10, 20, 30], r'Population size, $N=2n$')
S.save_panel(fig, 'b')
```
