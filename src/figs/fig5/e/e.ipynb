# Figure 5e
Probability of fixation by the sampling time given a mutant in a sample of one individual from island $1$ or from island $2$, for $N=20$, conditional on fixation. The sampling time is the expected fixation time at each $r$.

```python
import json

import numpy as np
import matplotlib.pyplot as plt

from fixloc import style as S

d = json.load(open('data.json'))
r, rst = np.array(d['r']), d['r_star']
```

```python
fig = plt.figure(figsize=(7.0, 6.2))
ax = fig.add_axes([0.075, 0.065, 0.36, 0.215])
ax.plot(r, d['p_fix_given_mutant_island1'], color=S.GREEN, label=r'Sample from island $1$')
ax.plot(r, d['p_fix_given_mutant_island2'], color=S.PURPLE, label=r'Sample from island $2$')
ax.axvline(rst, color=S.INK, lw=0.6, ls=(0, (1.5, 1.5)))
ax.text(rst + 0.006, 0.52, r'$r^*$', fontsize=8, va='bottom')
ax.set_xlim(1.0, 1.5)
ax.set_ylim(0.5, 1.0)
ax.set_xticks([1.0, 1.1, 1.2, 1.3, 1.4, 1.5])
ax.set_yticks([0.5, 0.6, 0.7, 0.8, 0.9, 1.0])
ax.set_xlabel(r'Mutant fitness, $r$')
ax.set_ylabel('Probability of fixation by the\nsampling time given a mutant')
S.legend(ax, loc='lower left')
S.save_panel(fig, 'e')
```
