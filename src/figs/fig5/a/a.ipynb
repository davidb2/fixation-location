# Figure 5a
A hand drawing of two islands with $n=10$. The notebook crops last-vertex-islands.png and places the drawing.

```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

from fixloc import style as S
```

```python
img = Image.open('last-vertex-islands.png').convert('RGB').crop((154, 79, 1498, 653))
fig = plt.figure(figsize=(7.0, 6.2))
ax = fig.add_axes([0.03, 0.70, 0.45, 0.27])
ax.imshow(np.asarray(img), interpolation='lanczos')
ax.set_axis_off()
ax.set_aspect('equal')
S.save_panel(fig, 'a')
```
