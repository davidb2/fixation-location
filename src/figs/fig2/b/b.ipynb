# Figure 2b
Temperature and fixation location probability on the path with $N=10$ vertices at $r=2$. The initial vertex is uniformly random.

```python
import json

import numpy as np
import matplotlib.pyplot as plt

from fixloc import style as S

d = json.load(open('data.json'))
edges = [tuple(e) for e in d['edges']]
```

```python
pos = {k: (-1.17 + 0.26 * k, 0.0) for k in range(10)}
```

```python
fig = plt.figure(figsize=(7.0, 4.75))
for y, vals, cmap, vmax in ((0.675, d['theta'], S.CM_TEMP, 2.5), (0.37, d['P_last'], S.CM_LAST, 0.5)):
  ax = fig.add_axes([0.285, y, 0.2, 0.29])
  fills = S.shade(cmap, dict(enumerate(np.array(vals))), 0, vmax)
  S.draw_graph(ax, pos, edges, fills, R=0.1, gap=0.025)
  ax.set_xlim(-1.3, 1.3)
  ax.set_ylim(-1.36, 1.3)
S.save_panel(fig, 'b')
```
