# Figure 4
Assembles the panels a to g. Panels a to d are crops of one hand drawing. The drawing holds the letters a to d.

```python
from fixloc import style as S

H_IN = 6.95
fig = S.assemble((7.0, H_IN), ['a/a.png', 'b/b.png', 'c/c.png', 'd/d.png', 'e/e.png', 'f/f.png', 'g/g.png'])
S.letter(fig, 0.004, 1 - 2.62 / H_IN - 0.03 + 0.022, 'e')
S.letter(fig, 0.004, 0.385, 'f')
S.letter(fig, 0.445, 0.385, 'g')
S.save_figure(fig, 'fig4')
```
