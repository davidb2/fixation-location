# Figure 4
Assembles the panels a to g and adds the panel letters.

```python
from fixloc import style as S

H_IN = 6.95
fig = S.assemble((7.0, H_IN), ['a/a.png', 'b/b.png', 'c/c.png', 'd/d.png', 'e/e.png', 'f/f.png', 'g/g.png'])
TOP, BOTTOM, ROW_GAP = 0.952, 0.672, 0.022
h = (TOP - BOTTOM - 2 * ROW_GAP) / 3
S.letter(fig, 0.004, 0.99, 'a')
for i, s in enumerate('bcd'):
  S.letter(fig, 0.352, TOP - i * (h + ROW_GAP) + 0.006, s)
S.letter(fig, 0.004, 1 - 2.62 / H_IN - 0.03 + 0.022, 'e')
S.letter(fig, 0.004, 0.385, 'f')
S.letter(fig, 0.445, 0.385, 'g')
S.save_figure(fig, 'fig4')
```
