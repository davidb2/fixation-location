# SI section 2: Fixation location, fixation probability, and fixation time
Each cell prints one sentence of the section and every number the sentence quotes.
The column SI is the quoted value and the column computed is the value at the precision of the SI.
Data: `python -m fixloc.make.si.factorization`. The tables with $N=9$ come from a cluster run of `fixloc.allgraphs`.

```python
import json

import numpy as np

from fixloc.make.si import quote

load = lambda f: json.load(open(f))
pct = '{:.0%}'
summ = load('allgraphs_summary.json')
sel = [e for e in summ if e['r'] in (1.25, 2.0, 10.0)]
neutral = [e for e in summ if e['r'] == 1.0]
```

```python
sizes = {e['N']: e['graphs'] for e in sel}
quote(
  r'Consider the $273{,}183$ connected undirected graphs with $N=5$ to $9$ vertices at $r=1.25$, $2$, and $10$.',
  ('273,183', sum(sizes.values()), '{:,}'),
)
print('graphs by N', sizes, '; regular graphs', {e['N']: e['regular_graphs'] for e in sel})
exc = {}
for e in sel:
  for x in e['exceptions']:
    exc.setdefault(x['g6'], []).append(dict(N=e['N'], r=e['r'], amp=round(x['amp'], 5), ratio_minus_1=round(x['gap'], 5)))
quote(r'The relative inverse temperature exceeds $1$ on every non-regular graph except four graphs.', ('4', len(exc), '{}'))
for g, v in exc.items():
  print(g, v)
dev = max(abs(x['amp'] - 1) for v in exc.values() for x in v)
quote(
  r'The four graphs have fixation probabilities within $0.6\%$ of the well-mixed value.',
  ('< 0.6%', dev, lambda x: '< 0.6%' if x < 0.006 else f'{x:.2%}'),
)
```

```python
n1 = sum(len(e['exceptions']) for e in neutral)
gap = max(-x['gap'] for e in neutral for x in e['exceptions'])
quote(
  r'Under neutral evolution the relative inverse temperature is below $1$ on $325$ of the $273{,}183$ graphs, by less than $3\%$.',
  ('325', n1, '{}'),
  ('273,183', sum(e['graphs'] for e in neutral), '{:,}'),
  ('< 3%', gap, lambda x: '< 3%' if x < 0.03 else f'{x:.2%}'),
)
print('exceptions by N:', {e['N']: len(e['exceptions']) for e in neutral})
print('the mean of 1/theta_v over the vertices is at least 1, so the relative shortfall is at most the absolute one,', round(gap, 5))
```

```python
d = {e['r']: e for e in load('directed_N5_summary.json')}
quote(
  r'Consider the $5{,}031$ strongly connected directed graphs with $N=5$ vertices and unequal temperatures.',
  ('5,031', d[1.0]['non_isothermal'], '{:,}'),
)
print(d[1.0]['graphs'], 'strongly connected directed graphs, of which', d[1.0]['isothermal'], 'have equal temperatures')
quote(
  r'The relative inverse temperature is below $1$ on $53\%$ of the graphs at $r=1$ and on $2\%$ at $r=10$.',
  ('53%', d[1.0]['frac_exceptions'], pct),
  ('2%', d[10.0]['frac_exceptions'], pct),
)
print({r: e['exceptions'] for r, e in d.items()})
```
