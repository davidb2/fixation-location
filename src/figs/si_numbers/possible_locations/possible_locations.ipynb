# SI section 4: Possible locations of fixation
Each cell prints one sentence of the section and every number the sentence quotes.
The column SI is the quoted value and the column computed is the value at the precision of the SI.
Data: `python -m fixloc.make.si.possible_locations` and the data of Figure S1a.

```python
import json

import numpy as np

from fixloc.make.si import quote

load = lambda f: json.load(open(f))
pct = '{:.0%}'
ex12 = load('../../figS1/a/data.json')
col = load('colonization_directed.json')
reach = load('impossible_reachability.json')
```

```python
zero = [v for v, p in zip(ex12['nodes'], ex12['P_last']) if p == 0]
quote(
  r'By \Cref{prop:gateway} the six vertices are impossible fixation locations, in agreement with the exact computation.',
  ('1, 2, 4, 7, 9, 11', zero, lambda x: ', '.join(map(str, x))),
)
print('the initial vertex is', ex12['initial'], '; the exact fixation location probabilities are zero at', zero)
```

```python
quote(
  r'An exhaustive search over mutant sets agrees with \Cref{prop:colonization-directed} on every directed graph with $3$ and $4$ vertices, from every initial vertex, and on $2{,}000$ random directed graphs with $5$ to $7$ vertices.',
  ('3, 4', col['sizes'], lambda x: ', '.join(map(str, x))),
  ('2,000', col['random_graphs'], '{:,}'),
  ('0', col['mismatches'], '{}'),
)
print(col['cases_checked'], 'pairs of a graph and an initial vertex, 2^6 x 3 + 2^12 x 4 + 2,000')
```

```python
quote(r'We checked reachability on all $986$ connected undirected graphs with $N=5$, $6$, and $7$ vertices.', ('986', reach['graphs'], '{}'))
f = reach['finite_r_graphs_with_impossible']
quote(
  r'At finite $r$ only the three paths and the three stars have impossible fixation locations.',
  ('3', f.get('path', 0), '{}'),
  ('3', f.get('star', 0), '{}'),
  ('6', sum(f.values()), '{}'),
)
quote(
  r'On the other $980$ graphs every vertex is a possible fixation location from every initial vertex.',
  ('980', reach['graphs'] - sum(f.values()), '{}'),
)
quote(
  r'Under colonization the impossible fixation locations are the initial vertex and the cut vertices on all $6{,}748$ pairs of a graph and an initial vertex, in agreement with \Cref{prop:colonization}.',
  ('6,748', reach['colonization_pairs_matching_theorem'], '{:,}'),
  ('0', len(reach['colonization_mismatches']), '{}'),
)
```
