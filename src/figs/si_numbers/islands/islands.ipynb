# SI section 8: Islands
Each cell prints one sentence of the section and every number the sentence quotes.
The column SI is the quoted value and the column computed is the value at the precision of the SI.
Data: `python -m fixloc.make.si.islands` and the data of Figure 5e.

```python
import json

import numpy as np

from fixloc.make.si import quote

load = lambda f: json.load(open(f))
pct = '{:.0%}'
from fixloc import islands

li = load('islands_last_island.json')
rows = {x['r']: x for x in li['rows']}
rs = load('r_star.json')
f5 = load('../../fig5/e/data.json')
R = (1.0, 1.26, 2.0, 5.0)
```

```python
quote(
  r'Take $n=10$, $m_{12}=10^{-3}$, and $m_{21}=10^{-1}$.',
  ('10', li['n'], '{}'),
  ('1e-03', li['m12'], '{:.0e}'),
  ('1e-01', li['m21'], '{:.0e}'),
)
quote(
  r'For a mutant starting on island $1$, the last wild type is on island $1$ with probability $0.75$, $0.50$, $0.21$, and $0.06$ at $r=1$, $1.26$, $2$, and $5$.',
  *[(si, rows[r]['p1_start_island1'], '{:.2f}') for r, si in zip(R, ('0.75', '0.50', '0.21', '0.06'))],
)
quote(
  r'For a mutant starting on island $2$, the last wild type is on island $1$ with probability $0.92$, $0.83$, $0.71$, and $0.70$ at the same fitnesses.',
  *[(si, rows[r]['p1_start_island2'], '{:.2f}') for r, si in zip(R, ('0.92', '0.83', '0.71', '0.70'))],
)
fac = lambda r: rows[r]['rho_from_island2'] / rows[r]['rho_from_island1']
quote(
  r'The fixation probability from island $2$ exceeds the fixation probability from island $1$ by a factor of $100$ at $r=1$.',
  ('100', fac(1.0), '{:.0f}'),
)
quote(r'The factor is $1.7$ at $r=2$.', ('1.7', fac(2.0), '{:.1f}'))
quote(
  r'With the initial vertex uniformly random and conditional on fixation, the last wild type is on island $1$ with probability $0.92$, $0.82$, $0.52$, and $0.39$ at $r=1$, $1.26$, $2$, and $5$.',
  *[(si, rows[r]['p1_uniform_initial_vertex'], '{:.2f}') for r, si in zip(R, ('0.92', '0.82', '0.52', '0.39'))],
)
quote(
  r'Island $1$ is the island of higher temperature.',
  ('True', li['temperature_island1'] > li['temperature_island2'], '{}'),
)
print('temperatures of the islands:', round(li['temperature_island1'], 3), round(li['temperature_island2'], 3))
quote(r'A tenth of the offspring of island $2$ are placed on island $1$.', ('0.1', li['m21'], '{:g}'))
```

```python
grid = np.linspace(1, 10, 9001)[1:]
lin = islands.p1_linear(grid)
quote(
  r'At the parameters of Figure~5\textbf{b,d} of the main text ($m_{12}=10^{-3}$, $m_{21}=10^{-1}$) the linearization gives $p_1<1/2$ for every $r>1$, an unbiased fitness $r^*=1$.',
  ('0.5', islands.p1_linear(1.0), '{:.1f}'),
  ('< 0.5', lin.max(), lambda x: '< 0.5' if x < 0.5 else f'{x:.4f}'),
)
print(f'largest p1 of the linearization on 1 < r <= 10: {lin.max():.6f}, at r = {grid[np.argmax(lin)]:.3f}')
quote(
  r'The exact Markov chain on the numbers of mutants on the two islands instead gives $r^*\approx 1.26$, $1.30$, and $1.30$ for $N=20$, $40$, and $80$.',
  *[(si, rs['N'][N]['r_star'], '{:.2f}') for N, si in (('20', '1.26'), ('40', '1.30'), ('80', '1.30'))],
)
print({N: round(v['r_star'], 4) for N, v in rs['N'].items()})
quote(
  r'The grid has $51$ values of $r$ from $1$ to $1.5$ (Figure~5 of the main text).',
  ('51', len(f5['r']), '{}'),
  ('1', min(f5['r']), '{:g}'),
  ('1.5', max(f5['r']), '{:g}'),
)
```
