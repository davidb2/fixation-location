# SI section 7: Testing a hypothesized population structure
Each cell prints one sentence of the section and every number the sentence quotes.
The column SI is the quoted value and the column computed is the value at the precision of the SI.
Data: `python -m fixloc.make.si.structure_test`. The exact tests on all graphs with $N=7$ vertices, the pair search, and the $87$ alternatives ran on a cluster.

```python
import json

import numpy as np

from fixloc.make.si import quote

load = lambda f: json.load(open(f))
pct = '{:.0%}'
from math import comb

from scipy.stats import norm

from fixloc import structure
from fixloc.make import fig4
from fixloc.make.si import structure_test as stt

sim = load('similar_classical_N7.json')
ex = {r: load(f'exact_test_r{r}_summary.json') for r in ('1', '2', '10')}
ratios = load('exact_test_ratios.json')
loc = {r: load(f'location_only_r{r}.json') for r in ('1', '2', '10')}
riv = load('river10.json')
ps = load('pair_search2.json')
f4 = load('fig4_pair.json')
alt = load('fig4_alternatives_summary.json')
```

## Similar fixation probability and fixation time, different fixation location

```python
e = sim['example']
quote(
  r'Among the $853$ connected graphs with $N=7$ vertices at $r=2$, $4{,}997$ pairs have fixation probabilities and fixation times within $1\%$ of each other.',
  ('853', sim['graphs'], '{}'),
  ('4,997', sim['close_pairs'], '{:,}'),
  ('1%', sim['tolerance'], pct),
)
quote(
  r'The distance is at most $0.44$ and exceeds $0.1$ on $19\%$ of the pairs.',
  ('0.44', sim['tv_max'], '{:.2f}'),
  ('19%', sim['share_tv_above_0p1'], pct),
)
quote(
  r'The most different pair has fixation probabilities $0.525$ and $0.526$ and fixation times $87.1$ and $87.4$ time steps.',
  ('0.525', e['rho'][0], '{:.3f}'),
  ('0.526', e['rho'][1], '{:.3f}'),
  ('87.1', e['T'][0], '{:.1f}'),
  ('87.4', e['T'][1], '{:.1f}'),
)
sig = lambda s: ', '.join(f'{x:.2f}' for x in s)
quote(
  r'The signatures in decreasing order are $(0.51, 0.40, 0.07, 0.01, \ldots)$ and $(0.26, 0.21, 0.21, 0.13, 0.13, \ldots)$.',
  ('0.51, 0.40, 0.07, 0.01', e['signatures'][0][:4], sig),
  ('0.26, 0.21, 0.21, 0.13, 0.13', e['signatures'][1][:5], sig),
)
edges = lambda E: ', '.join(f'{a}{b}' for a, b in E)
quote(
  r'The two graphs have edges $01, 04, 12, 14, 23, 36, 45$ and $01, 04, 14, 24, 34, 35, 36, 56$.',
  ('01, 04, 12, 14, 23, 36, 45', e['edges'][0], edges),
  ('01, 04, 14, 24, 34, 35, 36, 56', e['edges'][1], edges),
)
```

## Choosing between two structures

```python
quote(
  r'We simulated $4{,}000$ sequences of observations under each structure.',
  *[('4,000', ex[r]['M_per_truth'], '{:,}') for r in ex],
)
quote(
  r'We found $n$ for every connected graph with $N=7$ vertices against every connected graph obtained by adding or removing one edge, $17{,}294$ pairs in all.',
  *[('17,294', ex[r]['pairs'], '{:,}') for r in ex],
)
quote(
  r'Fixation locations alone need fewer observed fixation events than initial vertices and fixation times for $17\%$, $77\%$, and $91\%$ of the pairs at $r=1$, $2$, and $10$.',
  *[(si, ex[r]['share_location_beats_origin_time'], pct) for r, si in (('1', '17%'), ('2', '77%'), ('10', '91%'))],
)
quote(
  r'With the initial vertex observed, the fixation location needs fewer observed fixation events than the fixation time on $87\%$, $96\%$, and $95\%$ of the pairs.',
  *[(si, ex[r]['share_origin_location_beats_origin_time'], pct) for r, si in (('1', '87%'), ('2', '96%'), ('10', '95%'))],
)
quote(
  r'For every observed quantity the simulated count is close to $0.6$ times the bound on most pairs.',
  *[('0.6', v['median'], '{:.1f}') for r in ratios for v in ratios[r].values()],
)
for r, d in ratios.items():
  print(f'r = {r}:', {q: f"median {v['median']:.3f}, share from 0.5 to 0.7 {v['share_0p5_to_0p7']:.0%}" for q, v in d.items()})
print('censored searches:', {r: ex[r]['censored'] for r in ex})
```

```python
quote(
  r'The distribution $P_G$ separates all connected graphs with $N=7$ vertices up to relabeling, with three exceptions.',
  *[('3', loc[r]['graphs_tied'], '{}') for r in loc],
)
deg = lambda E: sorted({sum(v in e for e in E) for v in range(7)})
quote(
  r'The exceptions are the cycle, the complete graph, and the circulant graph where each vertex has an edge to each of the two nearest vertices on each side.',
  *[(si, len(E), '{} edges') for E, si in zip(loc['2']['uniform_law_edges'], ('7 edges', '14 edges', '21 edges'))],
)
print('degrees of the three graphs:', [deg(E) for E in loc['2']['uniform_law_edges']], '; graphs with the uniform distribution:', {r: loc[r]['graphs_uniform_law'] for r in loc})
```

```python
quote(
  r'We simulated $100{,}000$ sequences under each structure at $r=2$ with a uniformly random initial vertex.',
  ('100,000', stt.river10.__defaults__[1], '{:,}'),
  ('2', riv['r'], '{:g}'),
)
quote(r'The observer is right with probability $0.95$ after $47$ observed fixation events.', ('47', riv['n_exact'], '{:.0f}'))
quote(r'The bound would require $79$ events.', ('79', riv['n_bound'], '{:.0f}'))
```

## The pair of Figure 4e to g

```python
dash = lambda E: ', '.join(f'{a}-{b}' for a, b in E)
quote(
  r'The true structure has the edges $0$--$1$, $0$--$4$, $1$--$4$, $2$--$3$, $3$--$5$, $4$--$5$, and $5$--$6$.',
  ('0-1, 0-4, 1-4, 2-3, 3-5, 4-5, 5-6', fig4.TRUE, dash),
)
quote(
  r'The theorized structure has the edge $1$--$2$ in place of the edge $3$--$5$.',
  ('1-2', sorted(set(fig4.THEO) - set(fig4.TRUE)), dash),
  ('3-5', sorted(set(fig4.TRUE) - set(fig4.THEO)), dash),
)
quote(
  r'The search keeps the pairs with fixation probabilities and fixation times within $1\%$ at $r=2$, excluding pairs of isomorphic graphs, $4{,}732$ pairs in all.',
  ('4,732', ps['n_pairs'], '{:,}'),
)
top = ps['top'][0]
same = {frozenset(map(tuple, top['G'])), frozenset(map(tuple, top['H']))} == {frozenset(fig4.TRUE), frozenset(fig4.THEO)}
quote(r'The pair of Figure~4 has the largest rate of decrease among the $4{,}732$ pairs.', ('True', same, '{}'))
C = [x['C_full'] for x in ps['top']]
k = sum(abs(c - C[0]) < 1e-12 for c in C)
print(f'Chernoff information of the fixation location {C[0]:.4f}. The first {k} entries are this pair in other labelings and directions.')
print(f'The next pair has {C[k]:.4f}.')
quote(r'The monitored vertices are $2$ and $3$.', ('2, 3', f4['monitored'], lambda v: ', '.join(map(str, v))))
quote(
  r'Vertex $2$ is the monitored vertex last occupied by a wild type with probability $0.740$ on the true structure and $0.242$ on the theorized structure.',
  ('0.740', f4['P_true'][0], '{:.3f}'),
  ('0.242', f4['P_theorized'][0], '{:.3f}'),
)
quote(
  r'The probability first exceeds $0.95$ at $n=6$ and remains above $0.95$ for every $n\geq10$.',
  ('6', f4['first_n_095'], '{}'),
  ('10', f4['stays_above_095_from'], '{}'),
)
quote(
  r'The fixation probabilities of the two structures are $0.5264$ and $0.5242$.',
  ('0.5264', f4['rho'][0], '{:.4f}'),
  ('0.5242', f4['rho'][1], '{:.4f}'),
)
quote(r'The probability first reaches $0.95$ at $561{,}663$ trajectories.', ('561,663', f4['first_n_095_fixation'], '{:,}'))
quote(
  r'The fixation time distributions are exact, from the chain conditioned on fixation iterated for $4{,}000$ time steps, with all longer times as one category.',
  ('4,000', structure.time_pmf.__defaults__[0], '{:,}'),
)
quote(
  r'The five events of Figure~4\textbf{e} are the first five of $20$ fixation trajectories simulated on the true structure with the conditioned chain and random seed $11$.',
  ('20', structure.sample_events.__defaults__[0], '{}'),
  ('11', fig4.observable.__defaults__[0], '{}'),
)
```

## Choosing the monitored vertices

```python
chk = alt['check_fig4_true']
quote(
  r'We considered the $87$ connected structures on the seven vertices of Figure~4 with one edge added, removed, or moved relative to the theorized structure.',
  ('87', alt['alternatives'], '{}'),
)
quote(
  r'The number of observed fixation events for probability $0.95$ of choosing the true structure is approximately $2.706\,V/D^2$, from the normal approximation of the summed log likelihood ratio.',
  ('2.706', norm.ppf(0.95) ** 2, '{:.3f}'),
)
quote(
  r'For the pair of Figure~4 the approximation gives $9$ fixation events for vertices $2$ and $3$, against the exact $6$.',
  ('9', chk['pair_2_3'], '{:.0f}'),
  ('6', f4['first_n_095'], '{}'),
)
quote(
  r'For vertices $3$ and $6$ the approximation gives $1{,}273$ fixation events, against the exact $1{,}237$.',
  ('1,273', chk['pair_3_6'], '{:,.0f}'),
  ('1,237', f4['pair_3_6']['first_n_095'], '{:,}'),
)
rk = alt['pair']['ranking']
quote(
  r'The best pair is vertices $2$ and $3$, tied with vertices $5$ and $6$ by the symmetry of the theorized structure.',
  ('2, 3', alt['pair']['best'], lambda v: ', '.join(map(str, v))),
  ('5, 6', rk[1][0], lambda v: ', '.join(map(str, v))),
  ('35', rk[0][1], '{}'),
  ('35', rk[1][1], '{}'),
)
PH = np.array(f4['P_location'][1])
quote(
  r'Vertices $3$ and $6$ are the two likeliest fixation locations of the theorized structure.',
  ('3, 6', sorted(np.argsort(-PH)[:2].tolist()), lambda v: ', '.join(map(str, v))),
)
quote(r'The best triple is vertices $0$, $3$, and $6$.', ('0, 3, 6', alt['triple']['best'], lambda v: ', '.join(map(str, v))))
quote(
  r'Vertices $0$, $3$, and $6$ are the three likeliest fixation locations of the theorized structure.',
  ('0, 3, 6', sorted(np.argsort(-PH)[:3].tolist()), lambda v: ', '.join(map(str, v))),
)
print('fixation location probabilities of the theorized structure:', np.round(PH, 4).tolist())
```

```python
quote(
  r'Table~\ref{tab:identify} lists the shares of the $3{,}741$ pairs of alternatives distinguished within $100$ fixation events.',
  ('3,741', comb(alt['alternatives'], 2), '{:,}'),
)
quote(
  r'With two well-chosen vertices the observer detects $40\%$ of the alternatives within $100$ fixation events.',
  ('40%', alt['pair_2_3']['within_100'], pct),
)
quote(
  r'With two well-chosen vertices the observer distinguishes fewer than half of the pairs of alternatives.',
  ('< 50%', alt['identify']['best_pair'], lambda x: '< 50%' if x < 0.5 else f'{x:.0%}'),
)
quote(
  r'With every vertex monitored the observer distinguishes nearly all pairs.',
  ('98%', alt['identify']['all_sites'], pct),
)
```
