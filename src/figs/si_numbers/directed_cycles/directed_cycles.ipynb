# SI section 5: Directed cycles
Each cell prints one sentence of the section and every number the sentence quotes.
The column SI is the quoted value and the column computed is the value at the precision of the SI.
Data: `python -m fixloc.make.si.directed_cycles` and `fixloc.walk`.

```python
import json

import numpy as np

from fixloc.make.si import quote

load = lambda f: json.load(open(f))
pct = '{:.0%}'
from fixloc import walk

dev = {int(N): v['max_abs_N_pk_minus_1'] for N, v in load('directed_neutral_check.json').items()}
```

```python
rho = [(N, r, walk.exact_summary(N, r)['rho'], 1 / N if r == 1 else (1 - 1 / r) / (1 - r**-N)) for N in (10, 40) for r in (1.0, 2.0, 3.0)]
quote(
  r'The fixation probability $\rho$ equals $(1-r^{-1})/(1-r^{-N})$ for $r>1$ and $1/N$ for $r=1$.',
  ('< 1e-12', max(abs(a - b) for _, _, a, b in rho), lambda x: '< 1e-12' if x < 1e-12 else f'{x:.1e}'),
)
```

```python
quote(
  r'Exact computation gives the largest deviation $\max_k|Np_k-1|$ as $0.21$, $2.1\times10^{-2}$, $1.0\times10^{-3}$, $1.3\times10^{-5}$, and $2.5\times10^{-8}$ for $N=10$, $20$, $40$, $80$, and $160$.',
  ('0.21', dev[10], '{:.2f}'),
  ('2.1e-02', dev[20], '{:.1e}'),
  ('1.0e-03', dev[40], '{:.1e}'),
  ('1.3e-05', dev[80], '{:.1e}'),
  ('2.5e-08', dev[160], '{:.1e}'),
)
approx = {N: 4 * np.sqrt(2 * np.pi * N) * np.exp(-np.sqrt(np.pi * N)) for N in dev}
quote(r'An asymptotic expansion of the terms $\ell=1$ and $\ell=N-1$ gives $\max_k|Np_k-1|\approx4\sqrt{2\pi N}\,e^{-\sqrt{\pi N}}$.')
err = {N: approx[N] / dev[N] - 1 for N in dev}
quote(
  r'The approximation is within $10\%$ of the exact values for $N\ge80$.',
  *[('< 10%', abs(err[N]), lambda x: '< 10%' if x < 0.1 else f'{x:.2%}') for N in (80, 160)],
)
print({N: f'{approx[N]:.3e} against {dev[N]:.3e}, relative error {err[N]:+.2%}' for N in dev})
```

```python
l = np.arange(1, 10**6 + 1)
C = float((np.sqrt(l) * np.exp(1 - np.sqrt(l))).sum())
quote(r'with $C\coloneqq\sum_{\ell\ge1}\sqrt{\ell}\,e^{1-\sqrt{\ell}}<10.6$.', ('< 10.6', C, lambda x: '< 10.6' if x < 10.6 else f'{x:.3f}'))
print('C =', round(C, 4), '(terms up to l = 10^6; the remaining terms are below 10^-400)')
```

```python
m = lambda N, r: (N - 1) * r / (r - 1) - (r + 1) / (r - 1) ** 2
ok = all(abs((1 + m(N, 3) - (N / 2 - 1.5)) % N) < 1e-9 for N in range(4, 200))
quote(r'For $r=3$, vertex $1+m_N$ is vertex $N/2-3/2$ modulo $N$.', ('True', ok, '{}'))
```
