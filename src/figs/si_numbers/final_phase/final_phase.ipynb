# SI section 3: Fixation location and the final phase of fixation
Each cell prints one sentence of the section and every number the sentence quotes.
The column SI is the quoted value and the column computed is the value at the precision of the SI.
Data: `python -m fixloc.make.si.final_phase`, with the tables of the factorization section.

```python
import json

import numpy as np

from fixloc.make.si import quote

load = lambda f: json.load(open(f))
pct = '{:.0%}'
fp = load('final_phase_N8.json')
amp = load('amplifiers.json')
allg = load('../factorization/allgraphs_summary.json')
```

```python
quote(
  r'Consider all $11{,}117$ connected graphs with $N=8$ vertices at $r=2$, with a uniformly random initial vertex.',
  ('11,117', fp['graphs'], '{:,}'),
)
quote(
  r'The final phase is within $2.5\%$ of the value of $T^{\mathrm{end}}(u)$ with every $h(V\setminus\{v\})$ set to $1$.',
  ('< 2.5%', 1 - fp['ratio_min'], lambda x: '< 2.5%' if x < 0.025 else f'{x:.2%}'),
)
print('final phase divided by the value with every h set to 1: from', round(fp['ratio_min'], 4), 'to', round(fp['ratio_max'], 4))
```

## Table S1: the complete bipartite graphs with $100$ vertices at $r=2$

```python
fmt = {50: ('{:.1f}', '{:.1f}'), 20: ('{:.1f}', '{:,.0f}'), 10: ('{:.1f}', '{:,.0f}'), 1: ('{:.0f}', '{:,.0f}')}
table = {50: ('1.00', '1.0', '99.5', '1,335'), 20: ('1.17', '3.6', '362', '3,186'), 10: ('1.31', '8.6', '856', '7,001')}
table[1] = ('1.48', '99', '9,850', '77,269')
for a in amp:
  si = table[a['m']]
  quote(
    f"Table S1, m = {a['m']}: amplification, mean 1/theta_v of the fixation location, final phase, fixation time",
    (si[0], a['amp'], '{:.2f}'),
    (si[1], a['inv_theta_last'], fmt[a['m']][0]),
    (si[2], a['Tend'], fmt[a['m']][1]),
    (si[3], a['T'], '{:,.0f}'),
  )
```

```python
inc = [all(np.diff([a[k] for a in amp]) > 0) for k in ('amp', 'inv_theta_last', 'Tend', 'T')]
quote(
  r'As $m$ decreases from $50$ to $1$, the amplification, the mean inverse temperature of the fixation location, the final phase, and the fixation time increase.',
  ('True', all(inc), '{}'),
)
W = (2 * 99 + 1) / 2
err = max(abs(a['Tend'] / (W * a['inv_theta_last']) - 1) for a in amp)
quote(
  r'The final phase equals $99.5$ times the mean inverse temperature of the fixation location to within $0.1\%$, as \Cref{prop:endgame} predicts with $h(V\setminus\{v\})$ close to $1$.',
  ('99.5', W, '{:.1f}'),
  ('< 0.1%', err, lambda x: '< 0.1%' if x < 0.001 else f'{x:.3%}'),
)
```

```python
sel = [e for e in allg if e['r'] in (1.25, 2.0, 10.0)]
ex = {}
for e in sel:
  for x in e['exceptions']:
    if x['amp'] > 1 + 1e-9:
      ex.setdefault(x['g6'], []).append((e['N'], e['r'], round(x['amp'], 4)))
quote(
  r'On the connected undirected graphs with $N=5$ to $9$ vertices at $r=1.25$, $2$, and $10$, every amplifier except one has a fixation location with a larger mean inverse temperature than a uniformly random vertex.',
  ('1', len(ex), '{}'),
)
quote(r'The exception is a weak amplifier with nine vertices.', ('9', sorted({N for v in ex.values() for N, _, _ in v}), lambda x: ', '.join(map(str, x))))
print('the exception and its amplification at each r:', ex)
cls = {(e['N'], e['r']): e['classes']['amplifier']['count'] for N in range(5, 10) for e in load(f'amp_location_N{N}.json')}
print('amplifiers by (N, r):', cls)
```
